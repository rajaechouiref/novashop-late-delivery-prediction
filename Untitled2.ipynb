A1. Why does the company need historical orders with known final outcomes to train a supervised classifier? In a supervised learning framework, a model learns by associating input features (the predictors) with a known outcome (the target label). Without historical data where the final delivery status (late or on-time) is already recorded and verified, the classifier has no training signal (ground truth) to adjust its parameters, learn patterns, or understand which feature combinations correlate with delivery delays.

A2. Which two columns could reveal the answer after the order is finished but must not be used to predict at acceptance time? The two problematic columns are:

actual_delivery_days: The actual number of days it took to deliver the order.
refund_issued: Whether a refund was issued to the customer.
Using these columns during training causes target leakage. At the moment of order acceptance, we cannot know how many days the delivery will actually take or if a refund will be issued. Including them creates an unrealistically high performance during testing (as shown by the 95% accuracy in Task 9's leaky model) but makes the model completely useless for real-time predictions on new incoming orders.

A3. Why is order_id valuable for data quality although it is not a suitable predictive feature? While order_id has no predictive power (since it is a unique identifier/serial code with no logical relationship to delivery delays), it is critical for data quality because:

It allows for robust data auditing, such as identifying and removing duplicate records (like the 15 duplicate order IDs found and removed in Task 2).
It ensures data integrity and traceability when joining other tables (e.g., customer profiles or product inventories).
It helps track, debug, and reference specific high-risk orders for operational workflows (such as flagged reviews).

```python
print("\n--- Task 2: Data Audit ---")
print("Missing values per column:\n", orders.isna().sum())
print("\nDuplicate order IDs:", orders.duplicated(subset=['order_id']).sum())
orders = orders.drop_duplicates(subset=['order_id']).copy()
print("Shape after deduplication:", orders.shape)
```

B3. Does a missing value mean the corresponding order has zero price? Justify. No. A missing value (NaN/Null) in order_value indicates that the data was not recorded, is missing, or is unknown. It does not imply a price of zero. This is supported by the fact that these are actual product orders with multiple items (items_count > 0). Replacing missing values with a median during data preprocessing (rather than setting them to zero) is the standard and correct way to handle this without biasing the model.

```python
print("\n--- Task 3: Train / Test Split ---")
orders['order_date'] = pd.to_datetime(orders['order_date'])
orders = orders.sort_values(['order_date', 'order_id']).reset_index(drop=True)

cut = int(len(orders) * 0.8)
train = orders.iloc[:cut]
test = orders.iloc[cut:]
print('Training samples:', len(train), '| Testing samples:', len(test))
```

```python
numeric = ['order_value', 'items_count', 'customer_tenure_months', 'promised_days']
categorical = ['shipping_method', 'region', 'stock_available']

features = numeric + categorical
target = 'late_delivery'
```

We found 15 duplicate order IDs in our audit.

C3. Give one reason why duplicate records can bias evaluation. Duplicate records can bias evaluation because if identical data points appear in both the training set and the test set, the model's test performance will be artificially inflated. The model is tested on data it has already seen and memorized during training, which leads to overly optimistic evaluation metrics (like accuracy or precision) that do not reflect how the model performs on truly unseen, real-world data.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_steps = Pipeline([
    ('fill_missing', SimpleImputer(strategy='median')),
    ('scale', StandardScaler())
])

categorical_steps = Pipeline([
    ('fill_missing', SimpleImputer(strategy='most_frequent')),
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])

preprocess = ColumnTransformer([
    ('numbers', numeric_steps, numeric),
    ('categories', categorical_steps, categorical)
])
```

```python
from sklearn.linear_model import LogisticRegression

# Définition des variables d'entraînement et de test
X_train = train[features]
y_train = train[target]
X_test = test[features]
y_test = test[target]

model = Pipeline([
    ('prepare', preprocess),
    ('predictor', LogisticRegression(max_iter=2000, random_state=42))
])
model.fit(X_train, y_train)
predicted = model.predict(X_test)
probability = model.predict_proba(X_test)[:, 1]
```

D1. How many distinct orders are in the training and test samples?

There are 720 distinct orders in the training sample.

D2. Why can preprocessing statistics computed on the entire file invalidate the test evaluation? Calculating preprocessing statistics (such as the median for missing value imputation or the mean and standard deviation for scaling) on the entire dataset before splitting causes data leakage (also known as train-test contamination). This allows information from the future/test set to influence the training data. As a result, the model's performance on the test set is artificially inflated, and the evaluation no longer represents how the model would perform on completely unseen, independent data.

D3. What kind of real-world use does the chronological split approximate? A chronological split (sorting by date and splitting) closely mimics how the model will be used in production. In a real-world scenario, you train a model on past historical data and use it to predict outcomes for future, incoming orders. It tests the model's ability to generalize over time and accounts for temporal trends, shifting customer behaviors, or changes in delivery patterns.

```python
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score

baseline_class = int(y_train.mode().iloc[0])
baseline_pred = [baseline_class] * len(y_test)

print("\n--- Task 7: Model Evaluation ---")
print('Baseline Accuracy: ', accuracy_score(y_test, baseline_pred))
print('Model Accuracy:    ', accuracy_score(y_test, predicted))
print('Precision:         ', precision_score(y_test, predicted))
print('Recall:            ', recall_score(y_test, predicted))
print('Confusion matrix:\n', confusion_matrix(y_test, predicted))
```

H1. What happens to the review workload when the threshold increases?
As the classification threshold increases, the review workload decreases. This happens because a higher probability threshold requires the model to be more confident before flagging an order for review, resulting in fewer orders meeting or exceeding that threshold.

At 0.35, the team must review 68 orders.
At 0.55, the team must review 34 orders.
At 0.75, the team must review 23 orders.
H2. If the review team can inspect at most 20 of the 180 test orders, which of the three thresholds is feasible?
None of the three evaluated thresholds are fully feasible under a strict limit of 20 orders, as the highest threshold of 0.75 still yields 23 orders for review. To meet this operational constraint, the team would need to:

Select only the top 20 highest-risk orders (e.g., using the risk ranking).
Or, set a slightly higher threshold (e.g., 0.80) to bring the volume down to 20 or fewer.
H3. Can the model itself prove that an intervention will reduce late deliveries? Explain
No. The machine learning model can only establish correlations (predictive associations) based on historical patterns; it cannot establish causality. To prove that a specific intervention (such as upgrading shipping or altering stock) actually reduces late deliveries, the company would need to run a randomized controlled trial (like an A/B test) or employ causal inference methodologies.

```python
print("\n--- Task 8: Operational Prioritization ---")
results = test[['order_id', 'order_date', 'order_value']].copy()
results['actual_late_delivery'] = y_test.values
results['predicted_risk'] = probability
results['priority'] = (results['predicted_risk'] >= 0.55).map(
    {True: 'Review', False: 'Routine'}
)
print("Top 10 highest-risk test orders:")
print(results.sort_values('predicted_risk', ascending=False).head(10))

print("\nWorkload by threshold:")
for threshold in [0.35, 0.55, 0.75]:
    n_review = (probability >= threshold).sum()
    print('Threshold:', threshold, '| Orders for review:', n_review)
```

```python
from sklearn.tree import DecisionTreeClassifier

print("\n--- Task 9: Target Leakage Demonstration ---")
leaky_tree = DecisionTreeClassifier(random_state=42, max_depth=4)
leaky_tree.fit(train[['promised_days', 'actual_delivery_days']], y_train)
leaky_pred = leaky_tree.predict(test[['promised_days', 'actual_delivery_days']])
print('Invalid test accuracy (Leaky Model):', accuracy_score(y_test, leaky_pred))
```

I2. Why could a very high score still imply an unusable prediction system?
A very high score can indicate target leakage (data leakage). When a model is trained using features that contain direct or indirect information about the outcome which wouldn't actually be known at the time of prediction, the accuracy during training and testing is artificially inflated. In a real-world scenario, since those leaky features are unavailable when a new order is accepted, the model's actual performance will plummet, making it completely unusable in production.

I3. List the fields that must be excluded when predicting before dispatch.
To prevent target leakage and build a valid real-time prediction system, you must exclude:

actual_delivery_days (only known after delivery is completed)
refund_issued (only decided and processed after an issue occurs post-dispatch)
Any other post-acceptance outcomes or downstream tracking metrics.
